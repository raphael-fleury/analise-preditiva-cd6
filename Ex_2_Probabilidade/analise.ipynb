**Contexto de negócio:**

O time de negócios de empresa de e-commerce do ramo de assinatura de vinhos estava interessado em encontrar oportunidades na jornada de compra de seus clientes no site.

O principal indicador para o time é a taxa de conversão, ou seja, dos clientes que acessam o site, o percentual que adquire a assinatura.

```python
import pandas as pd

file_name = 'Aplicação 2 - para nota.xlsx'

def load_sheet(sheet_name):
    return pd.read_excel(file_name, sheet_name=sheet_name)

df_classico = load_sheet('Base')
df_classico.info()
```

Com o apoio da base histórica a seguir, faça as seguintes análises:

**1. Com base nas informações contidas na aba "Base", estime a probabilidade futura de que um visitante do site assine o produto.**

```python
num_assinaturas_classico = df_classico[df_classico['payment_success_flag'] == 1].shape[0]
num_sessoes_classico = df_classico.shape[0]
taxa_sucesso = num_assinaturas_classico / num_sessoes_classico

print("Número de pagamentos bem-sucedidos:", num_assinaturas_classico)
print("Número total de sessões:", num_sessoes_classico)
print("Taxa de sucesso:", f"{taxa_sucesso:.2%}")
```

**2. Nas outras abas temos amostras aleatórias de diferentes tamanhos (10, 15, 30, 150) da mesma base de dados da aba "Base". Calcule probabilidade do item 1) para os diferentes tamanhos de base.**

```python
tamanhos = [10, 15, 30, 150]

for tamanho in tamanhos:
    df = load_sheet(f'AA - n{tamanho}')
    print(f"Analisando DataFrame com {tamanho} amostras", df.shape[0])
    num_assinaturas = df[df['payment_success_flag'] == 1].shape[0]
    num_sessoes = df.shape[0]
    taxa_sucesso = num_assinaturas / num_sessoes
    print("Número de pagamentos bem-sucedidos:", num_assinaturas)
    print("Número total de sessões:", num_sessoes)
    print("Taxa de sucesso:", f"{taxa_sucesso:.2%}\n")
```

**3. Avalie os resultados obtidos nos itens 1) e 2). Os resultados foram os mesmos? O que explica essa dinâmica?**

Não. Os resultados mudaram drasticamente, e conforme o número de amostras foi aumentando, o valor foi se aproximando da % do método clássico.

Isso acontece pois quanto menos amostras, além de menos precisão e fidelidade, temos mais limitação estatística.

Por exemplo: em um cenário com 2 amostras, nosso resultado seria obrigatoriamente 0%, 50% ou 100%. Seria impossível chegar próximo ao valor real de 28,74%. Conforme o número de amostras cresce, a chance de acerto aumenta.
